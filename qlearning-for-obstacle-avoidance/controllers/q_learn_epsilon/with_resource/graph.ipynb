```python
import csv
import os

def create_csv_files(directory="state_files", num_files=10):
    if not os.path.exists(directory):
        os.makedirs(directory)
    
    for i in range(1, num_files + 1):
        file_name = f"state_{i}.csv"
        file_path = os.path.join(directory, file_name)
        
        data = [
            ["Iteration number", "0:FORWARD", "1:BACKWARD", "2:stop", "3:LEFT"]
        ]

        with open(file_path, mode='w', newline='') as file:
            writer = csv.writer(file)
            writer.writerows(data)

def add_statewise_actions_to_files(q, count):
    directory = "state_files"
    
    for i in range(10):
        file_name = f"state_{i}.csv"
        file_path = os.path.join(directory, file_name)
        
        # Format the state values
        q_values_state = [f"{count}", f"{q[i][0]}", f"{q[i][1]}", f"{q[i][2]}", f"{q[i][3]}"]

        # Check if the file exists and append the data
        if os.path.exists(file_path):
            # Open the file in append mode and write the row
            with open(file_path, mode='a', newline='') as file:
                writer = csv.writer(file)
                writer.writerow(q_values_state)  # Use writerow for a single row

            print(f"Data added to {file_path}: {q_values_state}")
        else:
            print(f"File {file_path} does not exist.")

 
```

```python
import csv
import os

def plot_all_actions_on_one_graph(directory="state_files"):
    # Loop through the files in the directory
    for file_name in os.listdir(directory):
        if file_name.endswith(".csv"):  # Process only CSV files
            file_path = os.path.join(directory, file_name)

            # Initialize lists to store data from the file
            iterations = []
            forward = []
            backward = []
            stop = []
            left = []

            # Read the data from the CSV file
            with open(file_path, mode='r', newline='') as file:
                reader = csv.reader(file)
                next(reader)  # Skip the header row
                
                for row in reader:
                    iterations.append(float(row[0]))  # Iteration number
                    forward.append(float(row[1]))    # 0:FORWARD
                    backward.append(float(row[2]))   # 1:BACKWARD
                    stop.append(float(row[3]))       # 2:STOP
                    left.append(float(row[4]))       # 3:LEFT
            
            # Plot the data
            plt.figure(figsize=(10, 6))
            plt.plot(iterations, forward, label='0:FORWARD', marker='o', linestyle='-', color='blue')
            plt.plot(iterations, backward, label='1:BACKWARD', marker='o', linestyle='-', color='red')
            plt.plot(iterations, stop, label='2:STOP', marker='o', linestyle='-', color='green')
            plt.plot(iterations, left, label='3:LEFT', marker='o', linestyle='-', color='purple')

            plt.title(f"Action Graph for {file_name}")
            plt.xlabel("Iteration number")
            plt.ylabel("Action Values")
            plt.legend()
            plt.grid(True)
            
            # Save or show the plot
            plt.savefig(f"{file_path}_plot.png")  # Save the plot as PNG file
            plt.show()  # Display the plot
```

```python
plot_all_actions_on_one_graph()
```

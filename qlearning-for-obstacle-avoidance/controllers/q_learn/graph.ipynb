```python
import matplotlib.pyplot as plt
import numpy as np

# Read the data from the text file
cummulative_filename = "./without_resource/qsumfile.txt"  
c_reward = np.loadtxt(cummulative_filename)

# Assuming the first column is X and the second column is Y
x = c_reward[:1500, 0]
y = c_reward[:1500, 1]

# Plot the data
plt.figure(figsize=(8, 6))
plt.plot(x, y, marker='o', linestyle='-', color='b', label='Data Line')

# Customize the graph
plt.title("Graph from Text File Data")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()

# Show the plot
plt.show()
```
